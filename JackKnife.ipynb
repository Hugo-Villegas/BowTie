### Cargar archivos desde Google Drive

Para cargar archivos desde Google Drive, primero necesitas montar tu Drive en Colab. Esto te permitirá acceder a tus archivos como si estuvieran en el sistema de archivos local de Colab.

```python
from google.colab import drive
drive.mount('/content/drive')
```

```python
import os

# Define la ruta a tu carpeta dentro de Google Drive.
# Por ejemplo, si tienes una carpeta llamada 'Mi_Carpeta' en la raíz de tu Drive.
drive_folder_path = '/content/drive/My Drive/InformesTrabajo/JackKnife'

# Verifica si la carpeta existe
if os.path.exists(drive_folder_path):
    # Lista los archivos y subcarpetas dentro de la carpeta
    files_and_folders = os.listdir(drive_folder_path)
    print(f"Archivos y carpetas en '{drive_folder_path}':")
    for item in files_and_folders:
        print(item)
else:
    print(f"La carpeta '{drive_folder_path}' no existe. Por favor, verifica la ruta.")
```

### Carga y Exploración Inicial del Archivo CSV

```python
import pandas as pd

# Construye la ruta completa al archivo CSV
csv_file_path = os.path.join(drive_folder_path, 'DataCompEneJun2026v2.csv')

# Carga el archivo CSV en un DataFrame de pandas
try:
    df = pd.read_csv(csv_file_path)
    print(f"Archivo '{os.path.basename(csv_file_path)}' cargado exitosamente.")
except FileNotFoundError:
    print(f"Error: El archivo '{csv_file_path}' no se encontró. Por favor, verifica la ruta y el nombre del archivo.")
except Exception as e:
    print(f"Ocurrió un error al cargar el archivo CSV: {e}")
```

### Inspeccionando el Archivo CSV para Identificar el Delimitador

```python
# Lee las primeras 10 líneas del archivo CSV para inspeccionar su formato
try:
    with open(csv_file_path, 'r', encoding='utf-8') as f:
        for i, line in enumerate(f):
            print(line.strip())
            if i >= 9:  # Muestra las primeras 10 líneas
                break
except FileNotFoundError:
    print(f"Error: El archivo '{csv_file_path}' no se encontró.")
except Exception as e:
    print(f"Ocurrió un error al leer el archivo CSV: {e}")
```
